# Eksplorasi Data (EDA) Skin Lesion — ISIC 2018 Task 3

Notebook EDA pendamping pipeline klasifikasi (`isic2018_resnet_pipeline.ipynb`). Semua analisis **read-only** (tidak training), hasil visual disimpan ke folder `output_eda/`. Tujuh tahap:

1. **Distribusi Kelas** — seberapa timpang antar 7 kelas diagnosis, baik pada partisi rilis asli (train/validation/test) maupun setelah training dipecah jadi train+val.
2. **Analisis `lesion_id`** — 1 image ≠ 1 lesi independen; cek `image_id` vs `lesion_id`, gambar per lesi, dan risiko **data leakage** saat split (lesi tercuil antar partition).
3. **Visualisasi Setiap Kelas** — 2–3 gambar/kelas untuk memahami dasar pembeda:
   shape, color, texture, border, size, global structure.
4. **Resolusi & Aspect Ratio** — distribusi lebar/tinggi/aspect ratio; apakah resize 224×224 masuk akal (cost, detail lesi, receptive field, memori, batch size).
5. **Distribusi Warna** — histogram RGB, brightness, contrast, saturation; termasuk sanity-check kuat-lemahnya ColorJitter agar augmentasi tidak menghancurkan warna asli.
6. **Duplikat / Near-Duplikat** — md5 (eksak) + dhash (near) antar partition, karena train ≈ validation bikin evaluasi bias.
7. **Perbandingan Skenario Split** — jumlah **per kelas** pada (1) split per-image [skema sekarang], (2) split per-lesion, (3) k-fold grouped per-lesion, (4) k-fold pada training + validation resmi; plus implikasinya ke class weight, sampling, augmentasi, dan interpretasi F1 per kelas.

> Dataset mengikuti rilis resmi ISIC 2018 Challenge. Analisis `lesion_id` memakai `ISIC2018_Task3_Training_LesionGroupings.csv` (rilis resmi, otomatis terdeteksi di `/kaggle/input` atau folder `dataset/`); `HAM10000_metadata.csv` hanya fallback. Tanpa keduanya, tahap 2 tetap menampilkan penjelasan konsep lengkap dan tahap 7 otomatis melewati skenario 2–4.

## 1. Konfigurasi & Setup (path dataset + parameter EDA)

Path dataset **terdeteksi otomatis**: di Kaggle dari `/kaggle/input` (folder yang memuat `ISIC2018_Task3_Training_Input` dipilih otomatis), di lokal dari folder `dataset/`. Semua artefak EDA (CSV + PNG) disimpan ke `EDA_CFG['output_dir']`.

````python
# =========================================================================
# SECTION 1: KONFIGURASI & SETUP (path dataset + parameter EDA)
# =========================================================================
import os
import glob
import hashlib
import json
from pathlib import Path

EDA_CFG = {
    # ---- PATH DATASET (ikuti rilis resmi ISIC 2018 Task 3, seperti pipeline utama) ----
    # Folder dicari berdasarkan NAMA, jadi aman baik untuk struktur "normal" (semua
    # folder anak langsung data_dir) maupun struktur Kaggle yang bersarang:
    #   /kaggle/input/datasets/<user>/<dataset>/ISIC2018_Task3_Training_Input/
    #       ISIC2018_Task3_Training_Input/ ... ISIC2018_Task3_Training_LesionGroupings.csv
    "data_dir": "dataset",   # path relatif lokal; di Kaggle otomatis ke /kaggle/input
                               # (boleh juga diisi path lengkap Kaggle, mis.
                               # "/kaggle/input/datasets/muhffikkri/isic2018-task-3")
    "train_img_dir": "ISIC2018_Task3_Training_Input",
    "test_img_dir":  "ISIC2018_Task3_Test_Input",
    "val_img_dir":   "ISIC2018_Task3_Validation_Input",
    "train_gt_dir":  "ISIC2018_Task3_Training_GroundTruth",
    "test_gt_dir":   "ISIC2018_Task3_Test_GroundTruth",
    "val_gt_dir":    "ISIC2018_Task3_Validation_GroundTruth",

    # ---- LESION GROUPINGS (WAJIB untuk analisis lesion_id & split anti-leakage) ----
    # File rilis resmi ISIC 2018 Task 3: ISIC2018_Task3_Training_LesionGroupings.csv
    "lesion_groupings": "",       # path eksplisit ke file lesion groupings; '' = auto-cari
    "lesion_groupings_file": "ISIC2018_Task3_Training_LesionGroupings.csv",  # nama file yang dicari (bukan path penuh)

    # ---- HAM10000 metadata (opsional, fallback bila LesionGroupings tidak ada) ----
    # Biarkan '' = auto-cari file bernama HAM10000_metadata.csv di /kaggle/input dan di dataset/.
    "ham10000_metadata": "",

    # ---- OUTPUT ----
    "output_dir": "output_eda",

    # ---- PARAMETER ANALISIS ----
    "class_columns": ["MEL", "NV", "BCC", "AKIEC", "BKL", "DF", "VASC"],
    "num_classes": 7,
    "img_size": 224,              # ukuran input model yang direncanakan (untuk diskusi resize)
    "val_ratio": 0.15,            # proporsi validation dari training set (sama dengan pipeline utama)
    "k_folds": 5,                 # jumlah fold untuk pratinjau cross validation (pipeline utama)
    "combine_train_val": False,   # True = gabung training + validation resmi lalu split lagi (TIDAK dipakai pipeline:
                                # 193 gambar validation resmi tidak punya lesion_id sehingga tidak bisa
                                # dijamin bebas lesi silang). Default False = skema competition:
                                # validation resmi = public test, test resmi = private test.
    "random_state": 42,           # seed reproduksibilitas split & sampling
    "sample_size": 300,           # sampel gambar untuk EDA resolusi & aspect ratio
    "color_sample_size": 200,     # sampel gambar untuk EDA warna
    "near_dup_sample": 400,       # sampel gambar untuk deteksi near-duplicate (dhash)
    "full_hash": True,            # hash md5 SEMUA gambar (deteksi exact duplicate eksak); False = hanya sampel
    "ham_near_threshold": 8,      # jarak hamming maksimum dhash yang dianggap near-duplicate
}

# ---- Resolusi path (robust terhadap dataset Kaggle yang tersusun bersarang) --------
# Di Kaggle dataset ISIC 2018 Task 3 bisa tersusun seperti:
#   /kaggle/input/datasets/<user>/<dataset>/ISIC2018_Task3_Training_Input/
#       ISIC2018_Task3_Training_Input/  ISIC2018_Task3_Test_Input/
#       ISIC2018_Task3_Validation_Input/  ISIC2018_Task3_*_GroundTruth/
#       ISIC2018_Task3_Training_LesionGroupings.csv
# yaitu SEMUA folder & CSV berada satu level DI DALAM ISIC2018_Task3_Training_Input.
# Karena itu setiap folder dicari berdasarkan NAMA (bukan hanya anak langsung
# `data_dir`): anak langsung -> satu level di atas -> penelusuran rekursif.
# `data_dir` boleh diisi path lengkap Kaggle, path relatif lokal, atau dikosongkan.

def _search_roots(cfg):
    """Titik awal pencarian: `data_dir`, satu level di atasnya, lalu /kaggle/input."""
    roots, seen = [], set()
    for cand in (str(cfg.get("data_dir", "") or ""), "/kaggle/input"):
        cand = cand.rstrip("/\\")
        for r in (cand, os.path.dirname(cand) if cand else ""):
            # jangan pernah jadi root filesystem (akan memindai seluruh disk)
            if r and os.path.isdir(r) and r not in seen and os.path.dirname(r) != r:
                seen.add(r)
                roots.append(r)
    return roots


_IMG_EXT = (".jpg", ".jpeg", ".png")


def _dir_has_images(path):
    return any(glob.glob(os.path.join(path, "*" + e)) for e in _IMG_EXT)


def _dir_score(path, name):
    """Semakin tinggi = semakin mungkin folder yang benar-benar berisi gambar.
    Layout Kaggle bisa memuat 'container' dengan nama yang sama persis
    (ISIC2018_Task3_Training_Input/ISIC2018_Task3_Training_Input), jadi kandidat
    yang punya gambar LANGSUNG harus menang atas folder kosongnya."""
    score = 0
    if _dir_has_images(path):
        score += 4
    if os.path.isdir(os.path.join(path, name)):   # container: memuat folder bernama sama
        score += 1
    if glob.glob(os.path.join(path, "*.csv")):
        score += 1
    return score


def _find_by_name(roots, name, want_dir):
    """Cari folder/file berdasarkan nama: anak langsung tiap root dulu, lalu rekursif.
    Untuk folder, kandidat yang benar-benar berisi gambar dianggap lebih tepat.
    Folder penuh gambar (>200 file) dipangkas agar pencarian tetap cepat."""
    if not name:
        return None
    is_ok = os.path.isdir if want_dir else os.path.isfile
    cands = [os.path.join(r, name) for r in roots]     # `name` absolut dihormati os.path.join
    cands = [p for p in cands if is_ok(p)]
    if not want_dir:
        if cands:
            return cands[0]
    elif not any(_dir_score(p, name) >= 4 for p in cands):
        # kandidat langsung container (tanpa gambar) -> carikan yang benar-benar berisi gambar
        for r in roots:
            for dirpath, dirnames, filenames in os.walk(r):
                dirnames.sort()
                p = os.path.join(dirpath, name)
                if is_ok(p) and p not in cands:
                    cands.append(p)
                n_img = sum(1 for f in filenames if f.lower().endswith(_IMG_EXT))
                if n_img > 200:
                    dirnames[:] = []   # folder gambar: tidak ada metadata CSV di dalamnya
    if not cands:
        return None
    if not want_dir:
        return sorted(cands, key=lambda p: (p.count(os.sep), len(p)))[0]
    return sorted(cands, key=lambda p: (-_dir_score(p, name), p.count(os.sep), len(p)))[0]


def _resolve_dir(cfg, key, label):
    p = _find_by_name(_search_roots(cfg), cfg[key], True)
    assert p, (label + ": folder '" + cfg[key] + "' tidak ditemukan.\n"
               "  Dicari mulai dari: " + ", ".join(_search_roots(cfg))
               + "\n  Ubah '" + key + "' di CONFIG/EDA_CFG bila nama foldernya berbeda.")
    return p


def _resolve_csv(cfg, key, label):
    folder = _resolve_dir(cfg, key, label)
    hits = sorted(glob.glob(os.path.join(folder, "*.csv")))
    assert hits, label + ": tidak ada file .csv di " + folder
    return hits[0]


def _name_candidates(fname):
    """Nama file + variasinya (basename, ekstensi ganda) agar typo kecil tidak menggagalkan."""
    cands = [fname, os.path.basename(fname or "")]
    if (fname or "").lower().endswith(".csv.csv"):
        cands.append(fname[:-4])
    out = []
    for c in cands:
        if c and c not in out:
            out.append(c)
    return out


TRAIN_IMG_DIR = _resolve_dir(EDA_CFG, "train_img_dir", "Training input")
TEST_IMG_DIR  = _resolve_dir(EDA_CFG, "test_img_dir", "Test input")
VAL_IMG_DIR   = _resolve_dir(EDA_CFG, "val_img_dir", "Validation input")
TRAIN_GT_PATH = _resolve_csv(EDA_CFG, "train_gt_dir", "Training ground truth")
TEST_GT_PATH  = _resolve_csv(EDA_CFG, "test_gt_dir", "Test ground truth")
VAL_GT_PATH   = _resolve_csv(EDA_CFG, "val_gt_dir", "Validation ground truth")

# "Data root" = folder induk folder gambar training (bisa satu level di bawah `data_dir`)
DATA_DIR = os.path.dirname(TRAIN_IMG_DIR)

OUTPUT_DIR = Path(EDA_CFG["output_dir"])
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ---- Cari metadata lesion_id (anti-leakage) ----
# Urutan: path eksplisit -> data root -> `data_dir` & /kaggle/input rekursif
#         (menangani path bersarang /kaggle/input/datasets/<user>/<dataset>/).
def _find_meta_file(cfg, explicit, fname, roots_extra=()):
    if explicit and os.path.isfile(explicit):
        return explicit
    roots = list(roots_extra) + _search_roots(cfg)
    for name in _name_candidates(fname):
        p = _find_by_name(roots, name, False)
        if p:
            return p
    return None


def _find_lesion_groupings(cfg):
    return _find_meta_file(cfg, cfg.get("lesion_groupings"),
                           cfg.get("lesion_groupings_file")
                           or "ISIC2018_Task3_Training_LesionGroupings.csv",
                           roots_extra=(DATA_DIR,))


def _find_ham_meta(cfg):
    return _find_meta_file(cfg, cfg.get("ham10000_metadata"),
                           "HAM10000_metadata.csv", roots_extra=(DATA_DIR,))

LESION_GROUPS_PATH = _find_lesion_groupings(EDA_CFG)
HAM_META_PATH = None if LESION_GROUPS_PATH else _find_ham_meta(EDA_CFG)

LESION_SOURCE = ("LesionGroupings (rilis resmi)" if LESION_GROUPS_PATH
                  else ("HAM10000_metadata (fallback)" if HAM_META_PATH else "TIDAK ADA"))
CLASS_COLUMNS = EDA_CFG["class_columns"]

# ---- Guard: folder gambar kosong = path salah resolve, bukan data hilang ----
def _n_images(path):
    return len(glob.glob(os.path.join(path, "*.jpg")))


N_TRAIN_IMG = _n_images(TRAIN_IMG_DIR)
N_TEST_IMG  = _n_images(TEST_IMG_DIR)
N_VAL_IMG   = _n_images(VAL_IMG_DIR)

print("DATA_DIR      :", DATA_DIR)
print("OUTPUT_DIR    :", OUTPUT_DIR)
print("Train gambar  :", N_TRAIN_IMG)
print("Test gambar   :", N_TEST_IMG)
print("Validation    :", N_VAL_IMG)
print("Lesion groups :", LESION_GROUPS_PATH or "TIDAK DITEMUKAN")
print("HAM10000 meta :", HAM_META_PATH if HAM_META_PATH else "tidak dipakai")
print("Sumber lesion_id:", LESION_SOURCE)
for _label, _dir, _n in (("Training", TRAIN_IMG_DIR, N_TRAIN_IMG),
                         ("Test", TEST_IMG_DIR, N_TEST_IMG),
                         ("Validation", VAL_IMG_DIR, N_VAL_IMG)):
    if _n == 0:
        print("PERINGATAN: folder %s tidak berisi .jpg (n=0); tahap yang membaca gambar "
              "akan dilewati. Path yang dipakai: %s" % (_label, _dir))
````

## 2. Imports & Helper Umum

````python
# =========================================================================
# SECTION 2: IMPORTS + HELPERS
# =========================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance

plt.rcParams["figure.dpi"] = 100
np.random.seed(EDA_CFG["random_state"])

CLASS_COLORS = {"MEL": "#d62728", "NV": "#7f7f7f", "BCC": "#2ca02c",
                "AKIEC": "#9467bd", "BKL": "#c49b9a", "DF": "#e377c2", "VASC": "#17becf"}


def onehot_to_label(df_raw):
    df = df_raw.copy()
    df["dx"] = df[CLASS_COLUMNS].idxmax(axis=1)
    return df[["image", "dx"]]


def load_gt(path):
    return onehot_to_label(pd.read_csv(path))


def stable_sample(files, n, seed):
    # Pilih n file secara deterministik (tidak acak-acakan antar run).
    rng = np.random.RandomState(seed)
    idx = rng.choice(len(files), size=min(n, len(files)), replace=False)
    return [files[i] for i in sorted(idx)]


def dhash_pil(im):
    # Perceptual hash 64-bit: bandingkan kedekatan kecerahan piksel 9x8.
    gray = im.convert("L").resize((9, 8), Image.BILINEAR)
    px = list(gray.getdata())
    bits = 0
    for r in range(8):
        for c in range(8):
            bits = (bits << 1) | (1 if px[r * 9 + c] >= px[r * 9 + c + 1] else 0)
    return bits


def hamming(a, b):
    return bin(a ^ b).count("1")
````

## 3. EDA Tahap 1 — Distribusi Kelas

Hitung jumlah & persentase gambar per kelas, plus **rasio setiap kelas** terhadap kelas terkecil dan terbesar (bukan hanya minoritas). **Mengapa penting**: distribusi ini menjadi dasar penentuan `loss_weight`/balancing di pipeline. Dengan model berbasis CrossEntropy, kelas mayoritas (`NV`) bisa mendominasi gradien sedangkan kelas minoritas (`DF`, `VASC`) berisiko tidak terpelajari.

Cell ini membandingkan dua skenario sekaligus:

- **Skenario A — data SEBAGAI rilis asli**: hitung `train` / `validation` / `test` resmi apa adanya. Ini yang dipakai sebagai acuan skema competition ISIC 2018.
- **Skenario B — sebagian training dipecah** jadi `train + val` (stratified, `val_ratio`) sehingga terlihat seperti skema pipeline saat ini. `test` tetap test resmi.

Keduanya disimpan terpisah (`eda_class_dist.csv` untuk skenario A, `eda_class_dist_split.csv` untuk skenario B) supaya tidak tercampur saat interpretasi.

````python
# =========================================================================
# EDA TAHAP 1: Distribusi Kelas
#   Skenario A — data SEBAGAI RILIS ASLI: training / test / validation resmi
#   Skenario B — sebagian training di-split stratified menjadi train + val
#               (skenario pipeline sekarang & rencana k-fold)
# =========================================================================
train_df = load_gt(TRAIN_GT_PATH)
test_df  = load_gt(TEST_GT_PATH)
val_df   = load_gt(VAL_GT_PATH)


def class_stats(df, name):
    cnt = df["dx"].value_counts().reindex(CLASS_COLUMNS).fillna(0).astype(int)
    pct = (cnt / cnt.sum() * 100).round(2)
    return pd.DataFrame({"class": CLASS_COLUMNS, name + "_count": cnt.values,
                         name + "_pct": pct.values})


def split_counts(df):
    return df["dx"].value_counts().reindex(CLASS_COLUMNS).fillna(0).astype(int)


# ---- Skenario A: partisi sesuai rilis resmi ----
dist = class_stats(train_df, "train")
dist = dist.merge(class_stats(test_df, "test"), on="class")
dist = dist.merge(class_stats(val_df, "val_resmi"), on="class")
dist["total"] = dist[["train_count", "test_count", "val_resmi_count"]].sum(axis=1)
print("=== SKENARIO A — partisi SEBAGAI RILIS ASLI ===")
print("total: train %d | test %d | val_resmi %d"
      % (dist["train_count"].sum(), dist["test_count"].sum(), dist["val_resmi_count"].sum()))
print(dist.to_string(index=False))
dist.to_csv(OUTPUT_DIR / "eda_class_dist.csv", index=False)

# ---- Skenario B: stratified split pada training ----
try:
    from sklearn.model_selection import train_test_split
    tr_imgs, va_imgs = train_test_split(train_df, test_size=EDA_CFG["val_ratio"],
                                        stratify=train_df["dx"],
                                        random_state=EDA_CFG["random_state"])
    dist_split = pd.DataFrame({
        "train": split_counts(tr_imgs),
        "val": split_counts(va_imgs),
        "val_resmi": split_counts(val_df),
        "test": split_counts(test_df),
    })
    for c in ["train", "val", "val_resmi", "test"]:
        dist_split[c + "_pct"] = (dist_split[c] / dist_split[c].sum() * 100).round(2)
    print("\n=== SKENARIO B — stratified split pada training (val_ratio=%s) ==="
          % EDA_CFG["val_ratio"])
    print("total: train %d | val_split %d | test %d"
          % (dist_split["train"].sum(), dist_split["val"].sum(), dist_split["test"].sum()))
    print(dist_split.to_string())
    dist_split.to_csv(OUTPUT_DIR / "eda_class_dist_split.csv")
    print("\nDua kandidat validation untuk monitoring per-epoch:")
    print("  validation resmi : %4d gambar (DF=%d, VASC=%d, AKIEC=%d)"
          % (dist_split["val_resmi"].sum(), dist_split.loc["DF", "val_resmi"],
             dist_split.loc["VASC", "val_resmi"], dist_split.loc["AKIEC", "val_resmi"]))
    print("  validation split : %4d gambar (DF=%d, VASC=%d, AKIEC=%d)"
          % (dist_split["val"].sum(), dist_split.loc["DF", "val"],
             dist_split.loc["VASC", "val"], dist_split.loc["AKIEC", "val"]))
    print("  -> validation resmi terlalu kecil & timpang untuk memilih epoch terbaik /")
    print("     early stopping; validation split jauh lebih stabil (detail di EDA Tahap 7).")
except ImportError:
    tr_imgs = va_imgs = None
    dist_split = None
    print("\nscikit-learn tidak tersedia -> SKENARIO B dilewati (install scikit-learn).")

# ---- Figure: dua skenario berdampingan ----
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
w = 0.27
xs = np.arange(len(CLASS_COLUMNS))
axes[0].bar(xs - w, dist["train_count"], w, label="Training", color="steelblue")
axes[0].bar(xs, dist["test_count"], w, label="Test resmi", color="indianred")
axes[0].bar(xs + w, dist["val_resmi_count"], w, label="Validation resmi", color="seagreen")
axes[0].set_xticks(xs); axes[0].set_xticklabels(CLASS_COLUMNS)
axes[0].set_title("Skenario A — partisi rilis asli")
axes[0].set_ylabel("Jumlah gambar"); axes[0].grid(axis="y", alpha=0.3); axes[0].legend()
if dist_split is not None:
    axes[1].bar(xs - w, dist_split["train"], w, label="Train (split)", color="steelblue")
    axes[1].bar(xs, dist_split["val"], w,
                label="Val (split, %s)" % EDA_CFG["val_ratio"], color="orange")
    axes[1].bar(xs + w, dist_split["val_resmi"], w, label="Val resmi", color="seagreen")
axes[1].set_xticks(xs); axes[1].set_xticklabels(CLASS_COLUMNS)
axes[1].set_title("Skenario B — stratified split pada training")
axes[1].set_ylabel("Jumlah gambar"); axes[1].grid(axis="y", alpha=0.3); axes[1].legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "eda_class_dist.png", dpi=120)
plt.show()

nv = dist.loc[dist["class"] == "NV", "train_count"].iloc[0]
df_ = dist.loc[dist["class"] == "DF", "train_count"].iloc[0]
print("Rasio NV : DF =", round(nv / df_, 1), ": 1")

# Rasio SETIAP kelas (terhadap kelas terkecil & terbesar), sesuai urutan CLASS_COLUMNS.
min_cnt = dist["train_count"].min()
max_cnt = dist["train_count"].max()
ratio_tbl = dist[["class", "train_count", "train_pct"]].copy()
ratio_tbl["rasio_vs_terkecil"] = (dist["train_count"] / min_cnt).round(2)
ratio_tbl["rasio_vs_terbesar"] = (max_cnt / dist["train_count"]).round(2)
if dist_split is not None:
    ratio_tbl["val_split_count"] = dist_split["val"].values
    ratio_tbl["val_resmi_count"] = dist_split["val_resmi"].values
print("\nRasio tiap kelas (jumlah gambar dibanding kelas terkecil / terbesar):")
print(ratio_tbl.to_string(index=False))
ratio_tbl.to_csv(OUTPUT_DIR / "eda_class_ratio.csv", index=False)
print("Rincian jumlah per kelas per skenario tersimpan di output_eda/eda_class_dist*.csv")
````

## 4. EDA Tahap 2 — Analisis `lesion_id` (image_id vs lesion_id, risiko leakage)

**`HAM10000` tidak selalu berarti 1 image = 1 lesi independen.** Ada lesi yang difoto lebih dari satu kali sehingga punya beberapa `image_id` di bawah satu `lesion_id`:

```
lesion_id
   |--- image A
   |--- image B
   `--- image C
```

Konsekuensinya: jika pada split `image A -> train` dan `image B -> validation`, model bisa "mengintip" lesi yang sama di dua partition sekaligus → **data leakage** dan estimasi performa jadi **terlalu optimistis**. Implementasi HAM10000 modern secara eksplisit melakukan split berbasis `lesion_id` untuk mencegah hal ini.

Sumber `lesion_id` dipakai berurutan:
1. **`ISIC2018_Task3_Training_LesionGroupings.csv`** (rilis resmi ISIC 2018 Task 3) — kolom `image`, `lesion_id`, `diagnosis_confirm_type`; mencakup seluruh 10.015 gambar training. Path lokal `dataset/`, Kaggle `/kaggle/input/...` (pencarian rekursif, juga path bersarang).
2. **`HAM10000_metadata.csv`** sebagai fallback bila file di atas tidak ada.
3. Tanpa salah satu: blok ini hanya menjelaskan konsep, dan split anti-leakage otomatis dimatikan.

> **Catatan cakupan**: file ini hanya memuat **training**. `lesion_id` untuk 193 gambar validation resmi tidak tersedia, jadi kita **tidak bisa** memastikan apakah lesi yang sama muncul di training ↔ validation resmi. Karena itu, validation resmi diperlakukan sebagai partisi *public test* terpisah, bukan untuk di-split lagi.

## 6. EDA Tahap 4 — Resolusi & Aspect Ratio

Periksa distribusi lebar, tinggi, dan aspect ratio sebelum memutuskan resize. Data ISIC asli umumnya bervariasi antara beberapa ratus piksel sampai ~1000 px, dan foto dermatoskop cenderung ~1:1 (lingkaran view) tetapi banyak juga yang crop persegi.

**Apakah harus resize 224×224?** Tidak otomatis harus, tapi ada konsekuensi langsung:
- **Computational cost & memory**: aktivasi & minibatch tumbuh kuadrat thd ukuran input.
- **Detail lesi**: resize mengecilkan bisa menghilangkan tekstur halus / border bergerigi (penting utk MEL vs BKL).
- **Receptive field**: struktur ResNet dirancang utk input 224; di input lain, field efektif layer konvolusi berubah relatif thd ukuran objek.
- **Batch size**: input lebih besar memaksa batch kecil (stabilitas optimasi berubah).

Bila aspek ratio tidak 1:1, resize 224×224 akan mendistorsi — opsi umum: resize dengan pisah lurus (mempertahankan ratio) lalu center-crop / pad.

Jika rentang asli sempit dan hasil augmentasi jauh menyeberang rentang itu, kuatkan jitter; jika menggeser ekstrem ke area yang tidak wajar, lemahkan.

````python
# =========================================================================
# EDA TAHAP 5: RGB histogram + brightness/contrast/saturation (sampel)
# =========================================================================
sample_files = stable_sample(glob.glob(os.path.join(TRAIN_IMG_DIR, "*.jpg")),
                             EDA_CFG["color_sample_size"], 7)

if not sample_files:
    print("PERINGATAN: TRAIN_IMG_DIR tidak berisi .jpg (n=0); EDA Tahap 5 (warna) dilewati.")
else:
    means = np.zeros(len(sample_files))
    vals = {"R": [], "G": [], "B": [], "brightness": [], "contrast": [], "saturation": []}
    hist_acc = {c: None for c in "RGB"}

    for f in sample_files:
        im = Image.open(f).convert("RGB")
        arr = np.asarray(im).astype(np.float32)
        lum = arr @ np.array([0.299, 0.587, 0.114])
        hsv = np.asarray(im.convert("HSV")).astype(np.float32)
        vals["R"].append(arr[..., 0].mean()); vals["G"].append(arr[..., 1].mean())
        vals["B"].append(arr[..., 2].mean())
        vals["brightness"].append(lum.mean()); vals["contrast"].append(lum.std())
        vals["saturation"].append(hsv[..., 1].mean() / 255.0)
        small = arr[::4, ::4].astype(np.uint8)
        for ci, c in enumerate("RGB"):
            h, _ = np.histogram(small[..., ci].ravel(), bins=64, range=(0, 256))
            hist_acc[c] = h / h.sum() if hist_acc[c] is None else hist_acc[c] + h / h.sum()
        im.close()

    for c in "RGB":
        hist_acc[c] /= len(sample_files)

    color_stats = pd.DataFrame({
        "stat": ["mean_R", "mean_G", "mean_B", "brightness", "contrast", "saturation"],
        "mean":  [np.mean(vals["R"]), np.mean(vals["G"]), np.mean(vals["B"]),
                  np.mean(vals["brightness"]), np.mean(vals["contrast"]), np.mean(vals["saturation"])],
        "median": [np.median(vals["R"]), np.median(vals["G"]), np.median(vals["B"]),
                   np.median(vals["brightness"]), np.median(vals["contrast"]), np.median(vals["saturation"])],
        "std":   [np.std(vals["R"]), np.std(vals["G"]), np.std(vals["B"]),
                  np.std(vals["brightness"]), np.std(vals["contrast"]), np.std(vals["saturation"])],
    }).round(2)
    print(color_stats.to_string(index=False))
    color_stats.to_csv(OUTPUT_DIR / "eda_color_stats.csv", index=False)

    chan_col = {"R": "red", "G": "green", "B": "blue"}
    fig = plt.figure(figsize=(13, 6.5))
    gs = fig.add_gridspec(2, 3, height_ratios=[1, 1])
    ax0 = fig.add_subplot(gs[0, :])
    for c in "RGB":
        bin_c = np.linspace(4, 252, 64)
        ax0.plot(bin_c, hist_acc[c], color=chan_col[c], label=c)
    ax0.set_title("Rata-rata histogram RGB (sampel)")
    ax0.set_xlabel("intensitas"); ax0.set_ylabel("fraksi piksel"); ax0.legend(); ax0.grid(alpha=0.3)

    for i, (key, ttl) in enumerate([("brightness", "Brightness"), ("contrast", "Contrast (std lum)"),
                                    ("saturation", "Saturation (HSV)")]):
        axb = fig.add_subplot(gs[1, i])
        axb.hist(vals[key], bins=30, color="#9467bd", edgecolor="white")
        axb.set_title(ttl); axb.set_xlabel("nilai rata-rata per gambar")
    axb.set_ylabel("jumlah gambar")
    fig.tight_layout()
    fig.savefig(OUTPUT_DIR / "eda_color_hists.png", dpi=120)
    plt.show()
````

### 7.1 Sanity-check ColorJitter (berapa jauh augmentasi menggeser distribusi?)

Simulasi `ColorJitter(brightness=0.2, saturation=0.2, contrast=0.2)` memakai `PIL.ImageEnhance` (faktor acak per gambar, seed tetap). Bandingkan statistik asli vs setelah jitter: kita ingin jitter memperluas distribusi **secukupnya** — tidak terlalu sempit (overfit) dan tidak terlalu ekstrem (membuat warna tidak realistis).

````python
# =========================================================================
# EDA TAHAP 5b: Simulasi ColorJitter (PIL ImageEnhance) & bandingkan statistik
# =========================================================================
jb, js_, jc = 0.2, 0.2, 0.2
rng = np.random.RandomState(11)
orig, jitt = {"brightness": [], "contrast": [], "saturation": []},              {"brightness": [], "contrast": [], "saturation": []}
hue_shifted = []

if not sample_files:
    print("PERINGATAN: tidak ada gambar tersampel; EDA Tahap 5b (ColorJitter) dilewati.")
else:

    for f in sample_files:
        im = Image.open(f).convert("RGB")
        lum = np.asarray(im).astype(np.float32) @ np.array([0.299, 0.587, 0.114])
        sat0 = np.asarray(im.convert("HSV")).astype(np.float32)[..., 1].mean() / 255.0
        orig["brightness"].append(lum.mean())
        orig["contrast"].append(lum.std())
        orig["saturation"].append(sat0)

        i2 = ImageEnhance.Brightness(im).enhance(1 + rng.uniform(-jb, jb))
        i2 = ImageEnhance.Color(i2).enhance(1 + rng.uniform(-js_, js_))
        i2 = ImageEnhance.Contrast(i2).enhance(1 + rng.uniform(-jc, jc))
        lum2 = np.asarray(i2).astype(np.float32) @ np.array([0.299, 0.587, 0.114])
        sat2 = np.asarray(i2.convert("HSV")).astype(np.float32)[..., 1].mean() / 255.0
        jitt["brightness"].append(lum2.mean())
        jitt["contrast"].append(lum2.std())
        jitt["saturation"].append(sat2)
        im.close()

    rows = []
    for k in orig:
        rows.append({"stat": k,
                     "orig_mean": np.mean(orig[k]), "jitter_mean": np.mean(jitt[k]),
                     "orig_median": np.median(orig[k]), "jitter_median": np.median(jitt[k]),
                     "orig_std": np.std(orig[k]), "jitter_std": np.std(jitt[k])})
    jdf = pd.DataFrame(rows).round(3)
    print(jdf.to_string(index=False))
    jdf.to_csv(OUTPUT_DIR / "eda_jitter_shift.csv", index=False)

    fig, axs = plt.subplots(1, 3, figsize=(13, 4))
    for ax, k in zip(axs, ["brightness", "contrast", "saturation"]):
        ax.boxplot([orig[k], jitt[k]])
        ax.set_xticks([1, 2])
        ax.set_xticklabels(["asli", "jitter +-0.2"])
        ax.set_title(k)
        ax.grid(alpha=0.3)
    fig.tight_layout()
    fig.savefig(OUTPUT_DIR / "eda_jitter_shift.png", dpi=120)
    plt.show()

    print("Bacaan: 'jitter_std' yg jauh lebih besar dari 'orig_std' = augmentasi memperlebar",
          "distribusi; jika 'jitter_median' keluar dari rentang orig, warna tidak realistis.")
````

## 8. EDA Tahap 6 — Duplikat / Near-Duplikat

Selain `lesion_id`, perlu juga memeriksa gambar yang **sama persis** (md5) atau **hampir sama** (dhash, 64-bit) karena: `train ~= validation` membuat evaluasi **bias**. Deteksi dilakukan antar partition: training (bisa di-split lagi), test resmi, validation resmi.

- **Exact duplicate**: md5 file. Jika `full_hash=True` (default), seluruh gambar pada ketiga partition di-hash (bisa butuh 1–2 menit).
- **Near duplicate**: perbandingan `dhash` berpasangan pada sampel (`EDA_CFG['near_dup_sample']`) dengan ambang hamming `EDA_CFG['ham_near_threshold']`.

````python
# =========================================================================
# EDA TAHAP 6: Exact duplicate (md5) seluruh partition + near dup (dhash)
# =========================================================================

# ---- 6a. exact duplicate ----
def file_md5(path, chunk=1 << 20):
    h = hashlib.md5()
    with open(path, "rb") as fh:
        for blk in iter(lambda: fh.read(chunk), b""):
            h.update(blk)
    return h.hexdigest()

print("Hash md5 (exact duplicate)...")
hash_rows = []
for part, imgdir in [("train", TRAIN_IMG_DIR), ("test", TEST_IMG_DIR), ("val", VAL_IMG_DIR)]:
    fl = sorted(glob.glob(os.path.join(imgdir, "*.jpg")))
    if not EDA_CFG["full_hash"]:
        fl = stable_sample(fl, EDA_CFG["near_dup_sample"], 3)
    for f in fl:
        hash_rows.append({"image": os.path.basename(f), "split": part, "md5": file_md5(f)})
hdf = pd.DataFrame(hash_rows, columns=["image", "split", "md5"])
hdf.to_csv(OUTPUT_DIR / "eda_dup_exact.csv", index=False)

dups = hdf[hdf.duplicated("md5", keep=False)].sort_values(["md5", "split"])
print("Jumlah entry gambar duplikat eksak (md5 sama):", len(dups))
if len(dups):
    for md5v, g in dups.groupby("md5"):
        print("  ", md5v[:8], "->", list(zip(g["image"].tolist(), g["split"].tolist())))

tr_h = set(hdf.loc[hdf["split"] == "train", "md5"])
te_h = set(hdf.loc[hdf["split"] == "test", "md5"])
va_h = set(hdf.loc[hdf["split"] == "val", "md5"])
print("Duplikat eksak training vs test :", len(tr_h & te_h))
print("Duplikat eksak training vs val  :", len(tr_h & va_h))

# ---- 6b. near duplicate (dhash) pada sampel training ----
print("\nHitung dhash near-duplicate pada sampel", EDA_CFG["near_dup_sample"], "gambar...")
sam = stable_sample(glob.glob(os.path.join(TRAIN_IMG_DIR, "*.jpg")),
                    EDA_CFG["near_dup_sample"], 5)
sh = [(os.path.basename(f), dhash_pil(Image.open(f))) for f in sam]
pairs = []
for i in range(len(sh)):
    for j in range(i + 1, len(sh)):
        d = hamming(sh[i][1], sh[j][1])
        if d <= EDA_CFG["ham_near_threshold"]:
            pairs.append((sh[i][0], sh[j][0], d))
pairs.sort(key=lambda x: x[2])
near = pd.DataFrame(pairs, columns=["image_a", "image_b", "hamming"])
near.to_csv(OUTPUT_DIR / "eda_dup_near.csv", index=False)
print("Pasangan near-duplicate (hamming <= %d): %d" % (EDA_CFG["ham_near_threshold"], len(near)))
if len(near):
    print(near.head(20).to_string(index=False))
````

## 9. EDA Tahap 7 — Perbandingan Skenario Split (jumlah per kelas per partition)

Cell ini membandingkan **empat skenario split** yang jadi bahan keputusan untuk pipeline utama. Semua angka **per kelas**, bukan hanya total:

```
                 skenario 1   skenario 2   skenario 3   skenario 4
                 per-image    per-lesion   k-fold      holdout + val resmi
df       ...       ...          ...          ...          ...
vasc     ...       ...          ...          ...          ...   <- selalu paling sedikit
```

| Skenario | Cara split | Gol | Keterangan |
|---|---|---|---|
| **1** | stratified per-**image** (`val_ratio`) | membrana bocor | skema pipeline lama; lesi sama bisa ada di train & val (589 lesi) |
| **2** | stratified per-**lesion** (`val_ratio`) | anti-leakage | holdout murni; tiap lesi utuh di satu partition |
| **3** | **k-fold** grouped per-lesion pada training | anti-leakage | `EDA_CFG['k_folds']`; tiap gambar jadi validation tepat 1× → paling banyak data efektif |
| **4** | **holdout**: split training `val_ratio` + **validation resmi** digabung | anti-leakage (sebagian) | `split_scheme="holdout"` di pipeline; validation resmi tidak pernah dilatih, tapi lesinya tidak diketahui |

Skenario 3 dan 4 adalah **dua skema split yang tersedia di pipeline** lewat `CONFIG['split_scheme']`, bukan dua tahap berurutan. Keduanya tidak memakai 193 gambar validation resmi untuk training.

`DF` dan `VASC` yang sangat jarang akan mempengaruhi:
- **class weight** (bobot CrossEntropy — pipeline memakai `inverse_frequency`),
- **sampling** (balancing/oversampling),
- **augmentation** (berapa kuat utk kelas minoritas),
- **interpretasi F1**: accuracy tinggi tapi F1 per kelas rendah = mayoritas mendominasi,
- **reliabilitas metrik per kelas**: dengan n sangat kecil, CI lebar & fluktuasi besar.

Skema evaluasi akhir yang dipilih pipeline lewat `CONFIG['split_scheme']`:
- `"kfold"` (default) → training di-split k-fold grouped per-lesi; validation resmi (193) = **public test** (tidak dilatih, tidak dipisah lagi), test resmi (1.512) = **private test**. Ini skema competition ISIC 2018.
- `"holdout"` → training di-split sekali `val_ratio`, lalu digabung dengan validation resmi (193) menjadi **validation set**; test resmi (1.512) = **private test**. Tidak ada public test karena 193 gambar sudah masuk validation, jadi angkanya tidak independen dari training.
- **tidak ada** opsi menggabungkan 193 gambar ke training (`combine_train_val`): file `LesionGroupings` tidak memuat `lesion_id` untuk partition tersebut sehingga lesi silang tidak bisa disingkirkan. Flag di EDA hanya dokumentasi, default `False`, dan tidak memengaruhi pipeline.

````python
# =========================================================================
# EDA TAHAP 7: Perbandingan skenario SPLIT (jumlah per kelas per partition)
#   1) stratified split per-image   -> skema pipeline lama (bocor lesi)
#   2) stratified split per-lesion  -> anti-leakage (holdout murni)
#   3) k-fold grouped per-lesion    -> pipeline: split_scheme='kfold' (default)
#   4) holdout: split val_ratio + validation resmi -> pipeline: split_scheme='holdout'
# 193 gambar validation resmi TIDAK punya lesion_id -> tidak pernah dipakai training
# =========================================================================
try:
    from sklearn.model_selection import (train_test_split, GroupKFold,
                                         StratifiedGroupKFold)
except ImportError:
    train_test_split = None

if train_test_split is None:
    print("scikit-learn tidak terinstall -> bagian split (Tahap 7) dilewati.")
    print("Install scikit-learn (Kaggle/Colab sudah tersedia) untuk menghitung train/val/test per kelas.")
else:
    HAS_LES = ("train_les" in dir()) and bool(train_les["lesion_id"].notna().any())

    def counts_of(df):
        return df["dx"].value_counts().reindex(CLASS_COLUMNS).fillna(0).astype(int)

    def add_pct(tbl, cols):
        for c in cols:
            tbl[c + "_pct"] = (tbl[c] / tbl[c].sum() * 100).round(1)

    # ---------- Skenario 1: stratified split per-IMAGE (pipeline sekarang) ----------
    tr_imgs, va_imgs = train_test_split(train_df, test_size=EDA_CFG["val_ratio"],
                                        stratify=train_df["dx"],
                                        random_state=EDA_CFG["random_state"])
    t1 = pd.DataFrame({"train": counts_of(tr_imgs), "val": counts_of(va_imgs),
                       "val_resmi": counts_of(val_df), "test": counts_of(test_df)})
    add_pct(t1, ["train", "val", "val_resmi", "test"])
    print("=== SKENARIO 1 — stratified split per-IMAGE (skema pipeline LAMA) ===")
    print("total: train %d | val %d | val_resmi %d | test %d"
          % (t1["train"].sum(), t1["val"].sum(), t1["val_resmi"].sum(), t1["test"].sum()))
    print(t1[["train", "train_pct", "val", "val_pct", "val_resmi", "test"]].to_string())
    t1.to_csv(OUTPUT_DIR / "eda_split_counts.csv")

    t1[["train", "val", "val_resmi", "test"]].plot(kind="bar", figsize=(10, 4.5),
                                                   rot=0, colormap="viridis")
    plt.title("EDA Tahap 7 — Skenario 1: jumlah gambar per kelas per partition")
    plt.xlabel("Kelas"); plt.ylabel("Jumlah gambar"); plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "eda_split_counts.png", dpi=120)
    plt.show()

    print("Kelas minoritas (DF, VASC) di validation split:",
          t1.loc["DF", "val"], t1.loc["VASC", "val"], "gambar")
    print("Kelas minoritas (DF, VASC) di validation resmi:",
          t1.loc["DF", "val_resmi"], t1.loc["VASC", "val_resmi"], "gambar")

    # ---------- Skenario 2: stratified split per-LESION (anti-leakage) ----------
    t2 = None
    if HAS_LES:
        les = train_les.dropna(subset=["lesion_id"])[["image", "dx", "lesion_id"]].copy()
        g = les.groupby("lesion_id").agg(dx=("dx", "first"), n_img=("dx", "size"))
        g_tr, g_va = train_test_split(g.reset_index(), test_size=EDA_CFG["val_ratio"],
                                      stratify=g["dx"], random_state=EDA_CFG["random_state"])
        tr_l, va_l = set(g_tr["lesion_id"]), set(g_va["lesion_id"])
        t2 = pd.DataFrame({"train": counts_of(les[les["lesion_id"].isin(tr_l)]),
                            "val": counts_of(les[les["lesion_id"].isin(va_l)])})
        add_pct(t2, ["train", "val"])
        print("\n=== SKENARIO 2 — stratified split per-LESION (anti-leakage) ===")
        print("lesion: train %d | val %d | overlap (harus 0): %d"
              % (len(tr_l), len(va_l), len(tr_l & va_l)))
        print("total gambar: train %d | val %d"
              % (t2["train"].sum(), t2["val"].sum()))
        print(t2.to_string())
        t2.to_csv(OUTPUT_DIR / "eda_split_lesion_counts.csv")

        # Ringkasan bocornya lesi pada kedua skenario
        les_map = les.set_index("image")["lesion_id"].to_dict()
        lk1 = {les_map[i] for i in tr_imgs["image"]} & {les_map[i] for i in va_imgs["image"]}
        leak_rows = les[les["lesion_id"].isin(lk1)]
        leak_tbl = pd.DataFrame([
            {"skenario": "1. split per-image", "lesion_bocor": len(lk1),
             "gambar_terlibat": int(len(leak_rows)),
             "pct_gambar": round(100.0 * len(leak_rows) / len(les), 1)},
            {"skenario": "2. split per-lesion", "lesion_bocor": len(tr_l & va_l),
             "gambar_terlibat": 0, "pct_gambar": 0.0},
        ])
        print("\n[RINGKASAN LEAKAGE] lesi yang muncul di train DAN val:")
        print(leak_tbl.to_string(index=False))
        leak_tbl.to_csv(OUTPUT_DIR / "eda_split_leak_summary.csv", index=False)
    else:
        print("\nSkenario 2-4 dilewati: sumber lesion_id tidak tersedia.")

    if t2 is not None:
        cmp_tbl = pd.DataFrame({"skenario_1_train": t1["train"],
                                "skenario_1_val": t1["val"],
                                "skenario_2_train": t2["train"],
                                "skenario_2_val": t2["val"]})
        cmp_tbl[["skenario_1_train", "skenario_1_val",
                "skenario_2_train", "skenario_2_val"]].plot(
            kind="bar", figsize=(11, 4.5), rot=0, colormap="Set2")
        plt.title("EDA Tahap 7 — split per-image vs per-lesion (jumlah per kelas)")
        plt.xlabel("Kelas"); plt.ylabel("Jumlah gambar"); plt.grid(axis="y", alpha=0.3)
        plt.tight_layout()
        plt.savefig(OUTPUT_DIR / "eda_split_compare.png", dpi=120)
        plt.show()

    # ---------- Skenario 3 & 4: k-fold grouped per-lesion ----------
    def kfold_report(pool, groups, tag, k):
        try:
            splitter = StratifiedGroupKFold(n_splits=k, shuffle=True,
                                            random_state=EDA_CFG["random_state"])
        except Exception:
            print("  (StratifiedGroupKFold tidak tersedia -> GroupKFold tanpa stratification)")
            splitter = GroupKFold(n_splits=k)
        rows = []
        for f, (a, b) in enumerate(splitter.split(pool, pool["dx"], groups)):
            r = {"skenario": tag, "fold": f, "n_train": len(a), "n_val": len(b),
                 "n_lesion_val": int(pd.Series(groups[b]).nunique())}
            vc = counts_of(pool.iloc[b])
            for c in CLASS_COLUMNS:
                r["val_" + c] = int(vc[c])
            rows.append(r)
        return pd.DataFrame(rows)

    if HAS_LES:
        k = EDA_CFG["k_folds"]
        print("\n=== SKENARIO 3 — k-fold=%d grouped per-lesion pada TRAINING saja ===" % k)
        rep3 = kfold_report(les, les["lesion_id"].values, "kfold_lesion_train", k)
        print(rep3.drop(columns=["skenario"]).to_string(index=False))
        print("Jumlah gambar per kelas di fold validation (min..max dari %d fold):" % k)
        for c in CLASS_COLUMNS:
            col = rep3["val_" + c]
            print("   %-5s %4d .. %4d" % (c, int(col.min()), int(col.max())))
        print("=> Tiap lesion masuk tepat satu fold; tiap gambar dipakai di validation 1x.")

        # ---------- Skenario 4: HOLDOUT (split val_ratio + validation resmi) ----------
        # Sama persis dengan CONFIG['split_scheme'] = "holdout" di pipeline: pakai
        # split g_tr/g_va (Skenario 2) lalu digabung dengan validation resmi.
        # Test resmi tetap private test. 193 gambar validation resmi tidak punya
        # lesion_id -> tiap gambar jadi grup, lesi silang dgn training tak bisa dicek.
        tr_h = les[les["lesion_id"].isin(set(g_tr["lesion_id"]))]
        va_h = pd.concat([les[les["lesion_id"].isin(set(g_va["lesion_id"]))], val_df],
                         ignore_index=True)
        t4 = pd.DataFrame({"train": counts_of(tr_h), "val": counts_of(va_h)})
        add_pct(t4, ["train", "val"])
        print("\n=== SKENARIO 4 — HOLDOUT: val_ratio + validation resmi (skema 'holdout' pipeline) ===")
        print("lesion split: train %d | val %d | overlap (harus 0): %d"
              % (len(g_tr), len(g_va), len(set(g_tr["lesion_id"]) & set(g_va["lesion_id"]))))
        print("total gambar: train %d | val %d (= %d split + %d validation resmi)"
              % (len(tr_h), len(va_h), t2["val"].sum(), len(val_df)))
        print("tidak ada public test: 193 gambar sudah jadi bagian validation set.")
        print(t4.to_string())
        t4.to_csv(OUTPUT_DIR / "eda_split_holdout_counts.csv")
        print("CATATAN: lesi silang antara training dan 193 gambar validation resmi TIDAK bisa")
        print("         disingkirkan karena validation resmi tidak punya lesion_id.")
        print("         Karena itu skema k-fold (Skenario 3) lebih aman: 193 gambar jadi")
        print("         public test dan tidak pernah dipakai training sama sekali.")

        rep3.to_csv(OUTPUT_DIR / "eda_cv_folds.csv", index=False)
        print("Detail fold k-fold (Skenario 3) -> output_eda/eda_cv_folds.csv")

        print("\n=== SKEMA EVALUASI (ikut CONFIG['split_scheme'] di pipeline) ===")
        if EDA_CFG["combine_train_val"]:
            print("combine_train_val=True  : training + validation resmi digabung, di-split k-fold.")
            print("                           TIDAK dipakai di pipeline: 193 gambar validation resmi tidak")
            print("                           punya lesion_id sehingga lesi silang tak bisa dicek.")
        else:
            print("combine_train_val=False (default): validation resmi = PUBLIC test,")
            print("                           test resmi = PRIVATE test -> skema competition ISIC 2018.")
            print("                           193 gambar validation resmi tidak pernah dipakai training.")
````

## 10. Ringkasan & Implikasi

Angka di bawah memakai data ISIC 2018 Task 3 (10.015 training / 193 validation / 1.512 test). Isi ulang bagian resolusi & warna setelah Tahap 4–5 dieksekusi.

1. **Ketimpangan kelas (rilis asli)**: training `NV` 6.705 (66,9%), `MEL` 1.113, `BKL` 1.099, `BCC` 514, `AKIEC` 327, `VASC` 142, `DF` 115. Rasio `NV`:`DF` ≈ **58:1** → pipeline wajib pakai `loss_weight`/balancing; F1 per kelas lebih informatif daripada accuracy. Validation resmi hanya 193 gambar dengan `DF` 1 dan `VASC` 3 → metrik per kelas di sana nyaris tidak bermakna, jangan dipakai sebagai satu-satunya acuan.
2. **`lesion_id` (Tahap 2)**: 10.015 gambar training **bukan** 10.015 lesi. Ada **7.470 lesi**, dan **1.956 lesi** punya >1 gambar (maks. 6) sehingga **4.501 gambar** ikut tersebar. Tidak ada lesi dengan label konflik, jadi pengelompokan lesi aman dipakai.
3. **Bukti leakage (Tahap 2 & 7)**: split stratified per-image 15% sekarang menyebarkan **589 lesi** ke train **dan** val (1.420 gambar, 14,2% data training). Artinya metrik validasi saat ini **bias optimistis** — model bisa saja hafal lesion yang sama, bukan generalize ke lesi baru.
4. **Skenario split (Tahap 7)**: per-lesi holdout menghasilkan 6.349 lesi train / 1.121 lesi val dengan **overlap 0**. k-fold grouped 5-fold memberi ~2.003 gambar validasi per fold dengan sebaran kelas sangat rata (mis. `DF` 23, `VASC` 28–29 per fold) dan tiap gambar dipakai validasi tepat sekali → **paling banyak data efektif tanpa leakage**. Keduanya sudah tersedia di pipeline lewat `CONFIG['split_scheme']`.
5. **Validation resmi 193 gambar**: tidak punya `lesion_id` sehingga **tidak boleh dipakai training** (lesi silang tak bisa disingkirkan). Pada `split_scheme="kfold"` (default) 193 gambar itu jadi **public test**; pada `split_scheme="holdout"` 193 gambar digabung ke validation set (bersama split `val_ratio`) dan test resmi jadi satu-satunya test. Opsi gabung-193-ke-training sengaja tidak disediakan.
6. **Visi per kelas**: catat pembeda visual yang paling mencolok per kelas (shape/color/texture/border/size).
7. **Resolusi**: mayoritas gambar sekitar X px, aspect ratio Y. Resize `img_size` pertimbangkan detail lesi vs memori/batch.
8. **Warna**: rentang brightness/contrast/saturation; hasil jitter ±0.2 menggeser statistik sebesar Z → pertimbangkan kekuatan ColorJitter.
9. **Duplikat**: ada/tidak exact & near duplicate, termasuk antar partition (jumlah pasangan). Bila ada, eksklusi dari split.

**Rekomendasi**: pakai **k-fold grouped per-lesion** (`CONFIG['split_scheme'] = "kfold"`, `k_folds`) sebagai skema utama; 193 gambar validation resmi diperlakukan sebagai public test dan tidak pernah dilatih. Gunakan `"holdout"` hanya bila ingin satu split besar + 193 gambar di validation untuk iterasi cepat.

Semua artefak tersimpan di `output_eda/` (CSV + PNG).
